```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler

def clean_and_preprocess_data(df, fill_strategy='mean'):
    cleaned_df = df.copy()

    if fill_strategy == 'mean':
        cleaned_df["ca"] = cleaned_df["ca"].fillna(cleaned_df["ca"].mean())
        cleaned_df["thal"] = cleaned_df["thal"].fillna(cleaned_df["thal"].mean())
    else:
        cleaned_df["ca"] = cleaned_df["ca"].fillna(cleaned_df["ca"].mode()[0])
        cleaned_df["thal"] = cleaned_df["thal"].fillna(cleaned_df["thal"].mode()[0])

    cleaned_df = cleaned_df[cleaned_df["chol"] < 400]

    new_targets = []
    for val in cleaned_df["target"]:
        if val > 0:
            new_targets.append(1)
        else:
            new_targets.append(0)
    cleaned_df["target"] = new_targets

    scaler = StandardScaler()
    num_cols = ["age", "trestbps", "chol", "thalach", "oldpeak"]
    cleaned_df[num_cols] = scaler.fit_transform(cleaned_df[num_cols])

    return cleaned_df
```

```python
column_names = [
    "age", "sex", "cp", "trestbps", "chol", "fbs", "restecg",
    "thalach", "exang", "oldpeak", "slope", "ca", "thal", "target"
]

raw_df = pd.read_csv("heart_data.csv", names=column_names, na_values="?")
cleaned_df = clean_and_preprocess_data(raw_df, fill_strategy='mean')

cleaned_df.to_csv("cleaned_heart.csv", index=False)
print("Notebook 1 Complete: 'cleaned_heart.csv' created.")
```
