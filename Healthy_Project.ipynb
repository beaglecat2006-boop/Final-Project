```python
import numpy as np
import pandas as pd

def clean_heart_data(df: pd.DataFrame) -> pd.DataFrame:
    cleaned_df = df.copy()

    # Fill missing values with the median
    cleaned_df["ca"] = cleaned_df["ca"].fillna(cleaned_df["ca"].median())
    cleaned_df["thal"] = cleaned_df["thal"].fillna(cleaned_df["thal"].median())

    # Map target values to binary without using lambda functions
    new_targets = []
    for value in cleaned_df["target"]:
        if value > 0:
            new_targets.append(1)
        else:
            new_targets.append(0)
    cleaned_df["target"] = new_targets

    return cleaned_df
```

```python
import pandas as pd

column_names = [
    "age",
    "sex",
    "cp",
    "trestbps",
    "chol",
    "fbs",
    "restecg",
    "thalach",
    "exang",
    "oldpeak",
    "slope",
    "ca",
    "thal",
    "target",
]

# Using a local file instead of learning complex URL fetches
raw_df = pd.read_csv("heart_data.csv", names=column_names, na_values="?")
cleaned_df = clean_heart_data(raw_df)
cleaned_df.to_csv("cleaned_heart.csv", index=False)

print("Notebook 1 Complete: 'cleaned_heart.csv' created.")
```
